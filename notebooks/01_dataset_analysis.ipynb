```python
import os

DATA_DIR = "../data/raw/chest_xray"

for split in ["train", "val", "test"]:
    print(f"\n{split.upper()}")

    for category in ["NORMAL", "PNEUMONIA"]:
        path = os.path.join(DATA_DIR, split, category)

        if os.path.exists(path):
            count = len(os.listdir(path))
            print(category, count)
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import os

normal_path = "../data/raw/chest_xray/train/NORMAL"

# Only files, not folders
files = [
    f for f in os.listdir(normal_path)
    if os.path.isfile(os.path.join(normal_path, f))
]

print("Images found:", len(files))
print(files[:5])

plt.figure(figsize=(15, 8))

for i, file in enumerate(files[:5]):

    image_path = os.path.join(normal_path, file)

    try:
        img = Image.open(image_path)

        plt.subplot(1, 5, i + 1)
        plt.imshow(img, cmap="gray")
        plt.axis("off")
        plt.title("NORMAL")

    except Exception as e:
        print("Could not open:", file)
        print(e)

plt.show()
```
